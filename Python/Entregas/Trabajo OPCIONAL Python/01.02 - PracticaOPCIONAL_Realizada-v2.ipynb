#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
convertidos = []
for n in valores:
    if n % 2 == 0 and n % 3 == 0:
        convertidos.append(f"par-mult3:{n}")
    elif n % 2 == 0:
        convertidos.append(f"par:{n}")
    elif n % 3 == 0:
        convertidos.append(f"mult3:{n}")
    else:
        convertidos.append(str(n))
print(convertidos)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
from pprint import pprint
resumen = {}
for e in estudiantes.values():
    promedio = sum(e["notas"]) / len(e["notas"])
    resumen[e["nombre"].upper()] = {
        "promedio": promedio,
        "aprobado": promedio >= 20,
        "max_min": (max(e["notas"]), min(e["notas"]))
    }
pprint(resumen, sort_dicts=False)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json
with open("notas_alumnos.txt", "w", encoding="utf-8") as f:
    f.write("Ana,10,20,30\nLuis,5,15\nMarta,100,50,25\n")
lista_alumnos = []
with open("notas_alumnos.txt", encoding="utf-8") as f:
    for linea in f:
        linea = linea.strip()
        if not linea:
            continue
        nombre, *notas = linea.split(",")
        notas = [int(n) for n in notas]
        promedio = sum(notas) / len(notas)
        lista_alumnos.append({
            "nombre": nombre.capitalize(),
            "promedio": promedio,
            "aprobado": promedio >= 20
        })
with open("resultado.json", "w", encoding="utf-8") as f:
    json.dump(lista_alumnos, f, ensure_ascii=False, indent=4)
with open("resultado.json", encoding="utf-8") as f:
    print(f.read())
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import csv
datos = {
    "1": {"nombre": "ana", "roles": ["admin", "user"]},
    "2": {"nombre": "luis", "roles": ["user"]},
    "3": {"nombre": "marta", "roles": ["moderator", "user"]}
}
with open("usuarios.json", "w", encoding="utf-8") as f:
    json.dump(datos, f, ensure_ascii=False, indent=4)
with open("usuarios.json", encoding="utf-8") as f:
    datos = json.load(f)
for d in datos.values():
    d["nombre"] = d["nombre"].upper()
with open("usuarios.csv", "w", newline="", encoding="utf-8") as f:
    escritor = csv.writer(f, lineterminator="\n")
    escritor.writerow(["id", "nombre", "roles"])
    for id_usuario, d in datos.items():
        escritor.writerow([id_usuario, d["nombre"], "|".join(sorted(d["roles"]))])
with open("usuarios.csv", encoding="utf-8") as f:
    print(f.read())
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
from collections import Counter
with open("fragmento-libro.txt", encoding="utf-8") as f:
    contenido = f.read().lower()
palabras_libro = re.findall(r"\w+", contenido)
top5 = dict(Counter(palabras_libro).most_common(5))
with open("frecuencias.json", "w", encoding="utf-8") as f:
    json.dump(top5, f, ensure_ascii=False, indent=4)
with open("frecuencias.json", encoding="utf-8") as f:
    print(f.read())
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
for i in range(1, 4):
    with open(f"ventas{i}.csv", "w", newline="", encoding="utf-8") as f:
        escritor = csv.writer(f, lineterminator="\n")
        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])
totales = {}
for i in range(1, 4):
    with open(f"ventas{i}.csv", newline="", encoding="utf-8") as f:
        for fila in csv.DictReader(f):
            totales[fila["producto"]] = totales.get(fila["producto"], 0) + int(fila["cantidad"])
with open("total_ventas.csv", "w", newline="", encoding="utf-8") as f:
    escritor = csv.writer(f, lineterminator="\n")
    escritor.writerow(["producto", "cantidad"])
    for producto, cantidad in totales.items():
        escritor.writerow([producto, cantidad])
with open("total_ventas.csv", encoding="utf-8") as f:
    print(f.read())
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
alumnos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]
with open("alumnos.json", "w", encoding="utf-8") as f:
    json.dump(alumnos, f, ensure_ascii=False)
with open("alumnos.json", encoding="utf-8") as f:
    alumnos = json.load(f)
estadisticas = {}
for a in alumnos:
    promedio = sum(a["notas"]) / len(a["notas"])
    estadisticas[a["nombre"]] = {
        "promedio": promedio,
        "max": max(a["notas"]),
        "min": min(a["notas"]),
        "aprobado": promedio >= 20
    }
print(json.dumps(estadisticas, indent=4))
```
