```python
"""
Arena de Mazmorras - Juego de combate por turnos en consola.
Creado por: Ismael Torres
"""

import random

AFIRMATIVO = {"si", "sí", "s", "yes", "y"}
NEGATIVO = {"no", "n"}

VIDA_JUGADOR = 150
VIDA_MONSTRUO = 60
DAÑO_JUGADOR = (30, 50)
DAÑO_MONSTRUO = (6, 14)

VIDA_JUGADOR_DIFICIL = 150
VIDA_MONSTRUO_DIFICIL = 240
DAÑO_JUGADOR_DIFICIL = (20, 34)
DAÑO_MONSTRUO_DIFICIL = (10, 24)

VIDA_JUGADOR_IMPOSIBLE = VIDA_MONSTRUO
VIDA_MONSTRUO_IMPOSIBLE = VIDA_JUGADOR
DAÑO_JUGADOR_IMPOSIBLE = DAÑO_MONSTRUO
DAÑO_MONSTRUO_IMPOSIBLE = DAÑO_JUGADOR

LINEA = "~" * 60


def preguntar_si_no(pregunta):
    while True:
        respuesta = input(pregunta).lower().strip()
        if respuesta in AFIRMATIVO:
            return True
        if respuesta in NEGATIVO:
            return False
        print("❌ . Escribe 'sí' o 'no', déjate de tonterias.")


def tirar_daño(rango):
    minimo, maximo = rango
    return random.randrange(minimo, maximo + 1, 2)


def pantalla_inicio():
    print("--- CREATED BY: Ismael Torres ---")
    print("👹 [Paseo Nocturno Uax Maker] 👹")
    print("¡Te damos la más cordial bienvenida a las catacumbas de la UAX! Respira hondo... lo vas a necesitar.\n")

    while True:
        if preguntar_si_no("¿Te atreves a desafiar el destino? "
                           "Escribe 'si' para entrar a la arena y 'no' si prefieres huir: "):
            print("\n¡Desafío aceptado! Estas por entrar en las mazmorras del edificio A...")
            return True

        if preguntar_si_no("\n🤨 ¿Seguro que quieres huir? ¡Aún estás a tiempo de cambiar de opinión! "
                           "('si' = huir / 'no' = quedarme): "):
            print("\nJAJAJA, me lo imaginaba de ti, cobarde como siempre. ¡Adiós!")
            return False

        print("\n😏 ¡Así me gusta! No te tenia fe pero veamos de que eres capaz. Vamos de nuevo...\n")


def preguntar_revancha():
    while True:
        if preguntar_si_no("\n🔁 ¿Quieres jugar otra partida? ('si' / 'no'): "):
            return True

        if preguntar_si_no("🤨 ¿Seguro que quieres abandonar la arena? "
                           "('si' = abandonar / 'no' = seguir jugando): "):
            return False

        print("\n🔥 ¡Esa es la actitud! La arena te estaba esperando...")


def pedir_accion(puede_bloquear):
    while True:
        print("¿Cuál será tu siguiente movimiento en el combate de la muerte?")
        print("1. Atacar con furia")
        if puede_bloquear:
            print("2. Bloquear y resistir")
        else:
            print("2. Bloquear y resistir (🔒 si bloqueas tendrás consecuencias: no disponible este turno)")

        accion = input("Elige opción (1 o 2): ").strip()

        if accion == "1":
            return accion
        if accion == "2":
            if puede_bloquear:
                return accion
            print("\n🔒 ¡No puedes bloquear dos turnos seguidos! Te quedaras 10 minutos más después de clases.\n")
        else:
            print("\n🌀 Opción no válida. Escribe 1 o 2.\n")


def combate():
    jugador = {"nombre": "biomedico pro", "vida": VIDA_JUGADOR}
    monstruo = {"nombre": "Marcello", "vida": VIDA_MONSTRUO}
    nombre_j = jugador["nombre"].capitalize()
    nombre_m = monstruo["nombre"].capitalize()
    bloqueo_previo = False

    print("\n⚔️ ¡Excelente! Forjando el destino y preparando la arena de combate...")
    print("\n--- 🛡️ MODO FÁCIL: EL BAUTIZO BIOMEDICO 🛡️ ---")
    print(f"Te adentras como {nombre_j} ({jugador['vida']} HP) frente a una masa viscosa "
          f"y amenazante: el {nombre_m} ({monstruo['vida']} HP).")
    print(f"\n🔥 ¡Preparate! Comienza el duelo mortal: {nombre_j} vs {nombre_m} 🔥\n")

    while True:
        print(LINEA)
        print(f"❤️ Tu salud: {jugador['vida']} HP  |  🟢 {nombre_m}: {monstruo['vida']} HP")

        accion = pedir_accion(puede_bloquear=not bloqueo_previo)
        bloqueando = accion == "2"
        bloqueo_previo = bloqueando

        if bloqueando:
            print("🛡️ Logras bloquear con tus apuntes de Anatomía Patológica, adoptando una postura de "
                  "contraataque defensivo inquebrantable.")
        else:
            daño = tirar_daño(DAÑO_JUGADOR)
            monstruo["vida"] = max(0, monstruo["vida"] - daño)
            print(f"💥 ¡ Le das un golpe en la cien a {monstruo['nombre']}! "
                  f"Le arrebatas {daño} puntos de vitalidad con un grito bestial.")

        if monstruo["vida"] <= 0:
            print("\n=== ¡VICTORIA! ===")
            print(f"¡Has derrotado a {monstruo['nombre']}!")
            print(f"Te han sobrado {jugador['vida']} HP de salud.")
            print("🎉 ¡Has demostrado tu valía en la arena!.")
            return True

        if bloqueando:
            print(f"✨ ¡Bloqueo magistral! {monstruo['nombre']} se estrella contra tu defensa "
                  f"y su impacto rebota sin dejar ni un rasguño.")
        else:
            daño = tirar_daño(DAÑO_MONSTRUO)
            jugador["vida"] = max(0, jugador["vida"] - daño)
            print(f"⚡ ¡ {monstruo['nombre']} se lanza y te muerde la nuca "
                  f"con un golpe seco de {daño} de daño!")

        if jugador["vida"] <= 0:
            print("\n=== GAME OVER ===")
            print(f"Has caído ante {monstruo['nombre']}. ¡Vaya mala suerte!")
            print("💀 Tu armadura queda cubierta de baba mientras la oscuridad te consume lentamente...")
            return False


def combate_dificil():
    jugador = {"nombre": "alumno todo plagio", "vida": VIDA_JUGADOR_DIFICIL}
    monstruo = {"nombre": "profesora Isabel", "vida": VIDA_MONSTRUO_DIFICIL}
    nombre_j = jugador["nombre"].title()
    nombre_m = monstruo["nombre"].title()
    bloqueo_previo = False

    print("\n📢 ¡El suelo tiembla! Tu victoria apenas te da un respiro: de las sombras sale alguien MUCHO peor...")
    print("\n--- 🔥 MODO DIFÍCIL: EL EXAMEN SORPRESA 🔥 ---")
    print(f"Te adentras como {nombre_j} ({jugador['vida']} HP) frente a la temida {monstruo['nombre']} "
          f"({monstruo['vida']} HP), con tres webs de Wikipedia copiadas en el bolsillo.")
    print(f"\n🔥 ¡Prepárate! Comienza el duelo mortal: {nombre_j} vs {nombre_m} 🔥\n")

    while True:
        print(LINEA)
        print(f"❤️ Tu salud: {jugador['vida']} HP  |  🟠 {nombre_m}: {monstruo['vida']} HP")

        accion = pedir_accion(puede_bloquear=not bloqueo_previo)
        bloqueando = accion == "2"
        bloqueo_previo = bloqueando

        if bloqueando:
            print("📚 Levantas un trabajo copiado de tres webs distintas como escudo, adoptando una postura "
                  "defensiva de dudosa ética académica.")
        else:
            daño = tirar_daño(DAÑO_JUGADOR_DIFICIL)
            monstruo["vida"] = max(0, monstruo["vida"] - daño)
            print(f"💥 ¡Le plantas un Ctrl+C, Ctrl+V en toda la cara a la {monstruo['nombre']}! "
                  f"Le arrebatas {daño} puntos de vitalidad con un plagio descarado.")

        if monstruo["vida"] <= 0:
            print("\n=== ¡VICTORIA! ===")
            print(f"¡Has derrotado a la {monstruo['nombre']}!")
            print(f"Te han sobrado {jugador['vida']} HP de salud.")
            print("🎉 ¡Por los pelos! Has demostrado tu valía en la arena.")
            return True

        if bloqueando:
            print(f"✨ ¡Bloqueo magistral! La {monstruo['nombre']} pasa tu escudo por el antiplagio "
                  f"y no encuentra nada, de lo bien que lo has parafraseado.")
        else:
            daño = tirar_daño(DAÑO_MONSTRUO_DIFICIL)
            jugador["vida"] = max(1, jugador["vida"] - daño)
            if jugador["vida"] == 1:
                print(f"😰 ¡La {monstruo['nombre']} te deja al borde del suspenso! "
                      f"Te quedas con 1 HP, pero todavía respiras.")
            else:
                print(f"⚡ ¡La {monstruo['nombre']} esta decepcionada de ti y te humilla delante de toda la clase "
                      f"y te quita {daño} de vida!")


def combate_imposible():
    jugador = {"nombre": "estudiante dispensa infinita", "vida": VIDA_JUGADOR_IMPOSIBLE}
    monstruo = {"nombre": "indomable Irene", "vida": VIDA_MONSTRUO_IMPOSIBLE}
    nombre_j = jugador["nombre"].title()
    nombre_m = monstruo["nombre"].title()
    bloqueo_previo = False

    print("\n💀 Un silencio helado recorre el pasillo... y entonces oyes unos pasos que nadie ha logrado detener jamás.")
    print("\n--- ☠️ MODO IMPOSIBLE: LA CONVOCATORIA EXTRADEMONIACA ☠️ ---")
    print(f"Te adentras como {nombre_j} ({jugador['vida']} HP) frente a la {monstruo['nombre']} "
          f"({monstruo['vida']} HP), que jamás ha concedido una prórroga.")
    print(f"\n🔥 ¡Que Dios te pille confesado! Comienza el duelo mortal: {nombre_j} vs {nombre_m} 🔥\n")

    while True:
        print(LINEA)
        print(f"❤️ Tu salud: {jugador['vida']} HP  |  🔴 {nombre_m}: {monstruo['vida']} HP")

        accion = pedir_accion(puede_bloquear=not bloqueo_previo)
        bloqueando = accion == "2"
        bloqueo_previo = bloqueando

        if bloqueando:
            print("📜 Despliegas tu dispensa infinita falsificada como un escudo de papel, adoptando una postura "
                  "defensiva de lo más burocrática.")
        else:
            daño = tirar_daño(DAÑO_JUGADOR_IMPOSIBLE)
            monstruo["vida"] = max(0, monstruo["vida"] - daño)
            print(f"💥 ¡Le tiras una dispensa arrugada a la cabeza de la {monstruo['nombre']}! "
                  f"Apenas le quitas {daño} puntos de vitalidad: ni se inmuta.")

        if monstruo["vida"] <= 0:
            print("\n=== ¡VICTORIA IMPOSIBLE! ===")
            print(f"¡Has derrotado a la {monstruo['nombre']}! Esto no debería haber pasado...")
            print(f"Te han sobrado {jugador['vida']} HP de salud.")
            print("🎉 ¡La dispensa infinita ha roto las leyes de la universidad!")
            return True

        if bloqueando:
            print(f"✨ ¡Bloqueo magistral! La profesora {monstruo['nombre']} pisa tu dispensa "
                  f"y, por una vez, el papel aguanta.")
        else:
            daño = tirar_daño(DAÑO_MONSTRUO_IMPOSIBLE)
            jugador["vida"] = max(0, jugador["vida"] - daño)
            print(f"⚡ ¡La profesora {monstruo['nombre']} te arrolla sin piedad con un golpe devastador de {daño} de daño!")

        if jugador["vida"] <= 0:
            print("\n=== GAME OVER ===")
            print(f"Has caído ante la {monstruo['nombre']}. ¡Era imposible, te lo advertimos!")
            print("💀 Tu dispensa se hace cenizas mientras la oscuridad te consume lentamente...")
            return False


def main():
    if not pantalla_inicio():
        return

    while True:
        combate()
        combate_dificil()
        combate_imposible()

        if not preguntar_revancha():
            break

    print("\n👋 Gracias por jugar. ¡Que los dioses de la arena te acompañen!")


if __name__ == "__main__":
    try:
        main()
    except (KeyboardInterrupt, EOFError):
        print("\n\n👋 Juego interrumpido. ¡Hasta la próxima, guerrero!")


```
